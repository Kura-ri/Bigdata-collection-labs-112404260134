```python
import requests

api_url = "https://115.159.159.82/api/devices?page=1&page_size=50" 

r = requests.get(api_url, params={"page": 1, "page_size": 50}, timeout=10)
print("status:", r.status_code)
print("url:", r.url)

data = r.json()
print(data.keys())
print(data)
```

```python
total = data["total"]       
page_size = data["page_size"]

total_pages = (total + page_size - 1) // page_size

print("total:", total)
print("page_size:", page_size)
print("total_pages:", total_pages)
```

```python
devices = []

for page in range(1, total_pages + 1):
    r = requests.get(
        api_url,
        params={"page": page, "page_size": page_size},   
        timeout=10
    )
    page_data = r.json()
    page_items = page_data["items"]                  
    devices.extend(page_items)

print("device count:", len(devices))
```

```python
from bs4 import BeautifulSoup

current_url = "https://115.159.159.82/companies"
r = requests.get(current_url, timeout=10)
soup = BeautifulSoup(r.text, "html.parser")

cards = soup.select("section.company-grid article.company-card")
print("本页企业数：", len(cards))
```

```python
card = cards[0]
print(card.prettify()[:1500])   # 先看清这条卡片的完整结构
```

```python
card = cards[0]

company_id = card.get("data-company-id")

# 名称：不挑标签，直接找文本里含「有限公司」的那个节点
name_node = card.find(string=lambda s: s and "有限公司" in s)
name = name_node.strip()

status = card.select_one("span.badge").get_text(strip=True)

details = {}
for item in card.select("dl > div"):
    details[item.select_one("dt").get_text(strip=True)] = \
        item.select_one("dd").get_text(strip=True)

industry = details["所属行业"]
zone = details["所在区域"]
device_count = int("".join(ch for ch in details["设备数量"] if ch.isdigit()))

print(company_id, "|", name, "|", status)
print(industry, zone, device_count)
```

```python
from urllib.parse import urljoin

companies = []
current = "https://115.159.159.82/companies"
next_selector = 'nav.pagination a:last-child'

for i in range(3):
    r = requests.get(current, timeout=10)
    soup = BeautifulSoup(r.text, "html.parser")

    page_cards = soup.select("section.company-grid article.company-card")
    for card in page_cards:
        details = {}
        for item in card.select("dl > div"):
            details[item.select_one("dt").get_text(strip=True)] = \
                item.select_one("dd").get_text(strip=True)

        name_node = card.find(string=lambda s: s and "有限公司" in s)
        companies.append({
            "company_id": card.get("data-company-id"),
            "name": name_node.strip(),
            "industry": details["所属行业"],
            "zone": details["所在区域"],
            "device_count": int("".join(ch for ch in details["设备数量"] if ch.isdigit())),
            "status": card.select_one("span.badge").get_text(strip=True),
        })

    print(f"第{i+1}页：{len(page_cards)}条")

    next_link = soup.select_one(next_selector)
    if not next_link or not next_link.get("href"):
        break
    current = urljoin(current, next_link.get("href"))

print("Company 总数：", len(companies))
```

```python
from collections import Counter

print("Company 总数：", len(companies))
print("唯一 company_id 数：", len(set(x["company_id"] for x in companies)))
print("Device 总数：", len(devices))
print("唯一 device_id 数：", len(set(x["device_id"] for x in devices)))

device_counts = Counter(x["company_id"] for x in devices)
mismatch_count = 0
for company in companies:
    cid = company["company_id"]
    expected = int(company["device_count"])
    actual = device_counts.get(cid, 0)
    if expected != actual:
        mismatch_count += 1

print("Company–Device 不一致企业数：", mismatch_count)
```

```python
from pathlib import Path
import json

data_dir = Path("data")
data_dir.mkdir(exist_ok=True)

with open(data_dir / "devices.json", "w", encoding="utf-8") as f:
    json.dump(devices, f, ensure_ascii=False, indent=2)

with open(data_dir / "companies.json", "w", encoding="utf-8") as f:
    json.dump(companies, f, ensure_ascii=False, indent=2)

print("数据已保存到 data/ 目录")
```
